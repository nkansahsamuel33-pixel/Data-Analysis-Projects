```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('Telco_customer_churnG3.xlsx')
df
```

```python
df.info()
```

```python
df.isna().sum()
```

# Handing missing values

```python
df.loc[df['Total Charges'].isna(), ['Tenure Months', 'Total Charges']]
```

```python
# Handling missing values
```

```python
df['Total Charges'] = df['Total Charges'].fillna(0)
```

# Grouping data into feature(x) and target(y)

```python
# Our feature data consist of categorical and numerical columns
cat_col = ['Internet Service','Online Security', 'Online Backup', 'Device Protection', 'Tech Support', 
          'Payment Method', 'Contract', 'Senior Citizen']

num_col = ['Tenure Months', 'Monthly Charges', 'Total Charges']

x = df[cat_col + num_col]
y = df['Churn Value']
```

# Train, Test, Split

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size = 0.20, random_state = 42, stratify = y)
```

#  Encoding

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

processor = ColumnTransformer( transformers = [('num', StandardScaler(), num_col),
                                ('cat', OneHotEncoder(drop = 'first', sparse_output = False), cat_col)] )

# p_xtrain = processed xtrain
# p_xtest = processed xtest

p_xtrain = processor.fit_transform(x_train)
p_xtest = processor.transform(x_test)
```

# LogisticRegression

```python
from sklearn.linear_model import LogisticRegression

# m1 = model 1
# creating m1 object from class LogisticRegression

m1 = LogisticRegression(max_iter = 1000,class_weight = 'balanced', random_state = 42)
# training the model
m1.fit(p_xtrain, y_train)

#  The model has learnt so its time to test it

test1 = m1.predict(p_xtest)
```

```python
# Metrics to evaluate the performance of the model

from sklearn.metrics import classification_report, confusion_matrix

m1_clreport = classification_report(y_test,test1)
m1_cfmatrix = confusion_matrix(y_test, test1)

print(' --- Classification Report ---')
print(m1_clreport)
print(' --- Confusion Matrix --- ')
print(m1_cfmatrix)
```

# Random Forest

```python
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier 

m2 = RandomForestClassifier(n_estimators = 100, class_weight = 'balanced', random_state = 42)
m2.fit(p_xtrain, y_train)
test2 = m2.predict(p_xtest)
```

```python
m2_clreport = classification_report(y_test, test2)
m2_cfmatrix = confusion_matrix(y_test, test2)

print(' --- Random Forest(RF) Classification Report ---')
print(m2_clreport)
print(' --- RF Confusion Matrix --- ')
print(m2_cfmatrix)
```

#  GradientBoosting

```python
m3 = GradientBoostingClassifier(n_estimators = 100, learning_rate = 0.1, random_state = 42)
m3.fit(p_xtrain, y_train)
test3 = m3.predict(p_xtest)
```

```python
m3_clreport = classification_report(y_test, test3)
m3_cfmatrix = confusion_matrix(y_test, test3)

print(' --- Gradient Boosting(GB) Classification Report ---')
print(m3_clreport)
print(' --- GB Confusion Matrix --- ')
print(m3_cfmatrix)
```
