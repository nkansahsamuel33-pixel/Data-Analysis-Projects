```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('dirty_cafe_sales (1).csv')
```

```python
df.head(20)
```

```python
df.info()
```

```python
# removing spacing from columns

df.columns = df.columns.str.replace(' ', '_')
```

```python
#check % of missing data

df.isna().mean() * 100
```

```python
#view the records where items are missing

df[df['Item'].isna()]
```

```python
#filling the missing data in items with Unknown, correcting casing and Error

df['Item'] = df['Item'].str.strip().str.title().fillna('Unknown').replace('Error', 'Unknown')
```

```python
df.head(10)
```

```python
# Fields where both Quantity and price per unit is missing

df[df['Quantity'].isna() & df['Price_Per_Unit'].isna()]
```

```python
# Fields where both Quantity and  Tototal amount is missing

df[df['Quantity'].isna() & df['Total_Spent'].isna()]
```

```python
#Dropping them
df = df.dropna(subset =['Quantity','Price_Per_Unit'], how ='all' )
```

```python
df =df.dropna(subset = ['Quantity', 'Total_Spent'], how = 'all')
```

```python
df.info()
```

```python
df[df['Quantity'] =='ERROR']
```

```python
# Replacing ERRORS and missing values with 0

df['Quantity'] = df['Quantity'].replace('ERROR',0).fillna(0)

df['Price_Per_Unit']=df['Price_Per_Unit'].replace('ERROR',0).fillna(0)

df['Total_Spent'] =df['Total_Spent'].replace('ERROR',0).fillna(0)
```

```python
df.tail(20)
```

### Chaning numeric columns datatype and calculations

```python
#Changed all numeric columns from string to numeric datatye

df['Quantity']= pd.to_numeric(df['Quantity'], errors = 'coerce')

df['Price_Per_Unit']= pd.to_numeric(df['Price_Per_Unit'], errors = 'coerce')

df['Total_Spent']= pd.to_numeric(df['Total_Spent'], errors = 'coerce')
```

```python
# Calculating missing numeric columns

df['Quantity'] = np.where(df['Quantity'] == 0.0, df['Total_Spent'] / df['Price_Per_Unit'], df['Quantity'])

df['Price_Per_Unit'] = np.where(df['Price_Per_Unit'] == 0.0, df['Total_Spent'] / df['Quantity'], df['Price_Per_Unit'])

df['Total_Spent'] = np.where(df['Total_Spent'] == 0.0, df['Quantity'] * df['Price_Per_Unit'], df['Total_Spent'])
```

```python
#Checking the distinct items in the column

df['Payment_Method'].unique()
```

```python
#fixing the payment method issues

df['Payment_Method'] = df['Payment_Method'].str.title().str.replace('ERROR','Unknown').fillna('Unknown').str.strip()
```

```python
#Checking the distinct items in the location column

df['Location'].unique()
```

```python
#fixing issues in the location column

df['Location'] = df['Location'].str.title().str.replace('ERROR','Unknown').fillna('Unknown').str.strip()
```

```python
#Viewing the null dates

df[df['Transaction_Date'].isnull()]
```

```python
#filling empty dates with unknown

df['Transaction_Date'] = df['Transaction_Date'].fillna('Unknown').str.replace('ERROR','Unknown')
```
